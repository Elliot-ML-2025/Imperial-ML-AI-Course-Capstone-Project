```python
#Import the Necessary Libraries
import numpy as np
import pandas as pd
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C
import matplotlib.pyplot as plt
```

```python
#Function 1

#Load the input and output data for function 1
inputs1 = np.load('initial_data/function_1/initial_inputs.npy')
outputs1 = np.load('initial_data/function_1/initial_outputs.npy')

input_data1 = pd.DataFrame(inputs1, columns = ['Input 1', ' Input 2'])
output_data1 = pd.DataFrame(outputs1, columns = ['Output'])
data1 = input_data1.join(output_data1)
data1
```

```python
kernel = C(1.0, constant_value_bounds="fixed") * RBF(length_scale = 0.1, length_scale_bounds="fixed")

gp = GaussianProcessRegressor(
    kernel=kernel,
    alpha=1e-10,
    normalize_y = True)

gp.fit(inputs1, outputs1)

```

```python
#Initialize grid for predictions 
grid_size = 1000
x1 = np.linspace(0, 1, grid_size)
x2 = np.linspace(0, 1, grid_size)
X1g, X2g = np.meshgrid(x1, x2)
grid = np.column_stack([X1g.ravel(), X2g.ravel()])

# GP predictions
mu, std = gp.predict(grid, return_std=True)

beta = 1.96   # UCB exploration parameter
UCB = mu + beta * std
```

```python
best_idx = np.argmax(UCB)
x_next = grid[best_idx]
x_next 
```
