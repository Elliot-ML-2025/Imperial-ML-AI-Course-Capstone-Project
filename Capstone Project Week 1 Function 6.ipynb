```python
import numpy as np
import pandas as pd
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C
import matplotlib.pyplot as plt
```

```python
#Function 6

#Load the input and output data for function 1
inputs = np.load('initial_data/function_6/initial_inputs.npy')
outputs = np.load('initial_data/function_6/initial_outputs.npy')

input_data = pd.DataFrame(inputs, columns = ['Input 1', ' Input 2', 'Input 3','Input 4',
                                             'Input 5'])
output_data = pd.DataFrame(outputs, columns = ['Output'])
data = input_data.join(output_data)
data
```

```python
kernel = C(1.0, constant_value_bounds="fixed") * RBF(length_scale = 0.1, length_scale_bounds="fixed")

gp = GaussianProcessRegressor(
    kernel=kernel,
    alpha=1e-10,
    normalize_y = True)

gp.fit(inputs, outputs)
```

```python
#Initialize the grid
grid_size = 23
x1 = np.linspace(0, 1, grid_size)
x2 = np.linspace(0, 1, grid_size)
x3 = np.linspace(0, 1, grid_size)
x4 = np.linspace(0, 1, grid_size)
x5 = np.linspace(0, 1, grid_size)
X1g, X2g, X3g, X4g, X5g = np.meshgrid(x1, x2, x3, x4, x5)
#Write the grid out in a list for the acquisition function to evaluate on
grid = np.column_stack([X1g.ravel(), X2g.ravel(), X3g.ravel(),X4g.ravel(), X5g.ravel()])


# GP predictions
mu, std = gp.predict(grid, return_std=True)

beta = 1.0   # UCB exploration parameter
UCB = mu + beta * std
```

```python
best_idx = np.argmax(UCB)
x_next = grid[best_idx]
x_next 
```
